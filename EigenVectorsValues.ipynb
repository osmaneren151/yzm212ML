```python
# YZM212 - Machine Learning Lab 3
# Eigenvalues & Eigenvectors Study
import numpy as np
# Matrix manipulation: verilerin matris formunda temsil edilmesi ve işlenmesi
# Eigenvalue: bir lineer dönüşümün belirli yönlerdeki ölçek katsayısı
# Eigenvector: dönüşüm sonrası yönü değişmeyen vektör
# np.linalg.eig(A) fonksiyonu:
# - kare matrisin özdeğerlerini ve özvektörlerini hesaplar
# - eigenvectors[:, i] -> eigenvalues[i] ile eşleşir
A = np.array([[4, 2],
              [1, 3]], dtype=float)
eigenvalues_np, eigenvectors_np = np.linalg.eig(A)
def manual_eigenvalues_2x2(matrix):
    a, b = matrix[0]
    c, d = matrix[1]
    trace = a + d
    det = a*d - b*c
    lambda1 = (trace + np.sqrt(trace**2 - 4*det)) / 2
    lambda2 = (trace - np.sqrt(trace**2 - 4*det)) / 2
    return np.array([lambda1, lambda2])
def manual_eigenvector_2x2(matrix, eigenvalue):
    a, b = matrix[0]
    c, d = matrix[1]
    v = np.array([b, eigenvalue - a], dtype=float)
    if np.allclose(v, 0):
        v = np.array([eigenvalue - d, c], dtype=float)
    if np.allclose(v, 0):
        v = np.array([1.0, 0.0])
    return v / np.linalg.norm(v)
eigenvalues_manual = manual_eigenvalues_2x2(A)
manual_vectors = [
    manual_eigenvector_2x2(A, eigenvalues_manual[0]),
    manual_eigenvector_2x2(A, eigenvalues_manual[1])
]
# KARŞILAŞTIRMA
comparison = np.allclose(np.sort(eigenvalues_np), np.sort(eigenvalues_manual))
# SONUÇ (TEK ÇIKTI)
print("Matrix A:\n", A)
print("\nEigenvalues (NumPy):", eigenvalues_np)
print("Eigenvalues (Manual):", eigenvalues_manual)
print("\nEigenvalues Match:", comparison)
```
